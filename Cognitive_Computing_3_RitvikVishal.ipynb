```python
import random

random.seed(1024160069)
numbers = [random.randint(100,900) for _ in range(100)]
print(numbers)
```

```python
count = 0
for i in range(len(numbers)):
  if numbers[i]%2 != 0:
    count+=1

print(count)
```

```python
count = 0
for i in range(len(numbers)):
  if numbers[i]%2 == 0:
    count+=1

print(count)
```

```python
k=[]

for i in range(len(numbers)):
  for j in range(2,int((numbers[i]/2)+1)):
    if numbers[i]%j == 0:
      break
  else:
    k.append(numbers[i])

print(k)
```

```python
del max
```

```python
dic = {}


for i in range(len(numbers)):
    if numbers[i] in dic:
        dic[numbers[i]] += 1
    else:
        dic[numbers[i]] = 1

print(max(dic, key=dic.get))
```

```python
rollno = 10241600
print(rollno)

A=set()
B=set()

for i in range(8):
  A.add((rollno % 10)*7)
  B.add((rollno % 10)*9)
  rollno = rollno//10


print(A)
print(B)
```

```python
print(A.union(B))

print(A.intersection(B))

print(A.difference(B))

print(B.difference(A))

print(A.symmetric_difference(B))
```

```python
print(A.issubset(B))

print(B.issuperset(A))
```

```python
my_dict.pop("city")

my_dict["location"] = "Kolakata"

my_dict
```

```python
my_dict["CGPA"] = 8.82

my_dict
```

```python
my_dict["age"] += 1

my_dict
```

```python
my_dict1 = dict()
my_dict1 = my_dict.copy()
my_dict1
```

```python
print(my_dict.pop("branch"))

del my_dict1["branch"]

# pop() removes and returns the deleted item, while del removes it without returning anything.

print(my_dict)
print(my_dict1)
```

```python
for x,y in my_dict.items():
  print(str(x)+"->"+str(y))
```

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key does not exist.")
```

```python
friend_dict = {
    "name": "Rahul",
    "age": 21,
    "branch": "ECE",
    "city": "Delhi",
    "college": "ABC College"
}

merged_dict = {**my_dict, **friend_dict}

print(merged_dict) # If both dictionaries have the same key, the value from the second dictionary (friend_dict) wins/overwrites the value from my_dict.
```
